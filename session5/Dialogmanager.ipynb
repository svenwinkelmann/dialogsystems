Ziele der heutigen Übung:

RASA (https://rasa.com/) ist ein Dialogmanagementsystem, welches es mittels Konfiguration ermöglicht einen text-basierten Chatbot zu erstellen. Hierfür definiert man mögliche Nutzeräußerungen im ```nlu.yml``` verschiedene Dialogabläufte in ```stories.yml``` und ```rules.yml``` und schließlich Systemantworten in ```domain.yml```.

RASA gab es früher mit dieser Funktionalität als Open-Source Variante. Diese lässt sich jedoch nur noch auf Python 3.10 installieren. Die neue Version von RASA ist zur Open-Source Variante kompatibel, ergänzt jedoch neben klassischen lernenden Modellen die Möglichkeit LLMs zu nutzen (werden wir in einer zukünftigen Übung einsetzen). Die neue Variante (RASA-Pro) benötigt einen Developer Account bei RASA und einen Lizenz-Key, welcher sich jedoch unkompliziert mit Angabe der EMail-Addresse anlegen lässt.

Wir installieren zuerst RASA-Pro in unserem virtuellen Environment:

````python
pip install rasa-pro
````

Wir werden heute hauptsächlich im Terminal arbeiten. Ein Terminal öffnest du in VSCode in der Menüleiste Terminal - Neues Terminal. Aktiviere dein virtuelle Environment mit:

```source .venv/bin/activate``` (Mac und Linux)

```.venv/Scripts/activate.ps1``` (Windows Powershell)

```.venv/Scripts/activate.bat``` (Windows CMD)

Um deinen Developer-Account bei RASA anzulegen, rufe folgende beiden Befehle auf:

```cd session5/rasa_nlu```

```rasa train```

Der Befehl sollte nach deiner eMail-Adresse fragen. An diese wird anschließend der Lizenz-Key geschickt. Bis zum Erhalt kannst du hier im Jupyter Notebook weiterarbeiten.

Sobald du den Lizenz-Key erhalten hast, musst du diesen in eine Umgebungsvariable im Terminal setzen:

```export RASA_LICENSE=<AddLicenseKeyHere>``` (Mac und Linux)

```set RASA_LICENSE=<AddLicenseKeyHere>``` (Windows)
