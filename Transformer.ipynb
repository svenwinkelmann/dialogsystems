## Beispiel Transformer Forward-Path

```python
import numpy as np
```

#### 1. Word Embeddings

vereinfachte Annahme: 3 Wörder mit einer Embedding-Dimension n=4

```python
DIM_EMB = 4
vocab = ["Tim", "isst", "Pizza"]
np.random.seed(42)
word_embeddings = np.random.randn(len(vocab), DIM_EMB)  # 3 Wörter, 4-dimensionale Embeddings zufällig generiert

print("Word embeddings für:")
for i in range(len(vocab)):
    print(f"{vocab[i]}: {word_embeddings[i]}")
```

#### 2. Positional Encoding

Als Funktionen nutzen wir:
1) sin(x)
2) sin(0.5x)
3) sin(0.25x)
4) sin(0.1x)

```python
from math import sin

pos_enc = []
for i in range(len(vocab)):
    emb_pos = [sin(i), sin(0.5*i), sin(0.25*i), sin(0.1*i)]
    emb_pos = np.array(emb_pos)
    pos_enc.append(emb_pos + word_embeddings[i].tolist())
    print(f"Embedding Position für Position {vocab[i]}: {emb_pos}")

pos_enc = np.array(pos_enc)
print("\npos_enc:", pos_enc)
```

#### 3. Attention

für unsere Gewichtsmatrizen Q, K, V nehmen wir Zufallswerte in einem richtigen System werden diese trainiert

```python
W_Q = np.random.randn(4, 4)
W_K = np.random.randn(4, 4)
W_V = np.random.randn(4, 4)
```

```python
# Query, Key, Value Berechnung für jedes Wort
Q = []
K = []
V = []
for i in range(len(vocab)):
    Q.append(pos_enc[i] @ W_Q)  # @ ist Matrixmultiplikation
    K.append(pos_enc[i] @ W_K)
    V.append(pos_enc[i] @ W_V)
```

```python
def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum()
```

```python
attention_outputs = []

for i in range(len(vocab)):
    sim_values = []
    for j in range(len(vocab)):
        sim_values.append(np.dot(Q[i], K[j]))
    sim_values = softmax(sim_values)
    attention_outputs.append(sim_values @ V)

attention_outputs = np.array(attention_outputs)
print("Attention Outputs:", attention_outputs)
```

#### 4. Residual Connection

```python
residual_output = attention_outputs + pos_enc
print("Residual Output:", residual_output)
```

#### Zusammenfassung

```python

print("Word Embeddings:\n", word_embeddings)
print("\nInput mit Positional Encoding:\n", pos_enc)
print("\nAttention Output:\n", attention_outputs)
print("\nOutput nach Residual Connection:\n", residual_output)
```
